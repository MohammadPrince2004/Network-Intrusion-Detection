# **Data**

| العمود              | المعنى                                                       | النوع                      |
| ------------------- | ------------------------------------------------------------ | -------------------------- |
| `dur`               | مدة الاتصال/التدفق بالثواني                                  | Numerical                  |
| `proto`             | بروتوكول الاتصال مثل TCP, UDP, ICMP                          | Categorical                |
| `service`           | الخدمة المستخدمة مثل HTTP, FTP, DNS                          | Categorical                |
| `state`             | حالة الاتصال مثل FIN, CON, INT                               | Categorical                |
| `spkts`             | عدد الـ Packets المرسلة من المصدر                            | Numerical                  |
| `dpkts`             | عدد الـ Packets المرسلة من الوجهة                            | Numerical                  |
| `sbytes`            | عدد الـ Bytes المرسلة من المصدر                              | Numerical                  |
| `dbytes`            | عدد الـ Bytes المرسلة من الوجهة                              | Numerical                  |
| `rate`              | معدل نقل البيانات/الحزم في الاتصال                           | Numerical                  |
| `sload`             | معدل تحميل البيانات من المصدر                                | Numerical                  |
| `dload`             | معدل تحميل البيانات من الوجهة                                | Numerical                  |
| `sloss`             | عدد الـ Packets المفقودة من المصدر                           | Numerical                  |
| `dloss`             | عدد الـ Packets المفقودة من الوجهة                           | Numerical                  |
| `sinpkt`            | متوسط الزمن بين Packets المصدر                               | Numerical                  |
| `dinpkt`            | متوسط الزمن بين Packets الوجهة                               | Numerical                  |
| `sjit`              | مقدار التذبذب (Jitter) في Packets المصدر                     | Numerical                  |
| `djit`              | مقدار التذبذب (Jitter) في Packets الوجهة                     | Numerical                  |
| `swin`              | حجم نافذة TCP للمصدر                                         | Numerical                  |
| `stcpb`             | رقم TCP sequence الأولي للمصدر                               | Numerical                  |
| `dtcpb`             | رقم TCP sequence الأولي للوجهة                               | Numerical                  |
| `dwin`              | حجم نافذة TCP للوجهة                                         | Numerical                  |
| `tcprtt`            | وقت Round Trip في اتصال TCP                                  | Numerical                  |
| `synack`            | الزمن بين SYN و SYN/ACK                                      | Numerical                  |
| `ackdat`            | الزمن بين SYN/ACK و ACK                                      | Numerical                  |
| `smean`             | متوسط حجم الحزم المرسلة من المصدر                            | Numerical                  |
| `dmean`             | متوسط حجم الحزم المرسلة من الوجهة                            | Numerical                  |
| `trans_depth`       | عمق المعاملة داخل بروتوكول HTTP                              | Numerical                  |
| `response_body_len` | حجم محتوى الـ Response                                       | Numerical                  |
| `ct_src_dport_ltm`  | عدد الاتصالات الحديثة من نفس المصدر إلى نفس Destination Port | Numerical                  |
| `ct_dst_sport_ltm`  | عدد الاتصالات الحديثة إلى نفس Source Port                    | Numerical                  |
| `is_ftp_login`      | هل حدث تسجيل دخول FTP أم لا                                  | Binary                     |
| `ct_ftp_cmd`        | عدد أوامر FTP في الاتصال                                     | Numerical                  |
| `ct_flw_http_mthd`  | عدد طرق HTTP المستخدمة مثل GET/POST                          | Numerical                  |
| `is_sm_ips_ports`   | هل المصدر والوجهة يستخدمان نفس IP/Port                       | Binary                     |
| `attack_cat`        | نوع الهجوم مثل DoS, Exploits, Fuzzers                        | Categorical / Target محتمل |
| `label`             | هل الاتصال طبيعي أم هجوم                                     | Binary / Target            |

# **Cleaning Data**

```python
train_data.duplicated().sum()
```

```python
train_data.drop_duplicates(inplace=True)
```

```python
train_data.reset_index(inplace=True , drop=True)
```

```python
train_data.isna().sum().sort_values(ascending=False).head()
```

```python
train_data.info()
```

# **Balancing Data**

```python
normal_train = train_data[train_data["label"]==0]
normal_train.reset_index(drop=True , inplace= True)
attacked_train = train_data[train_data["label"]==1]
attacked_train.reset_index(drop=True , inplace= True)
print(normal_train.shape)
print(attacked_train.shape)
```

```python
balanced_data =pd.concat([normal_train,attacked_train],axis=0)
balanced_data.head()
```

```python
shuffled_balanced_data = balanced_data.sample(frac=1.0, random_state=42).reset_index(drop=True)
shuffled_balanced_data.head()
```

```python
shuffled_balanced_data.to_csv("balnced_data.csv",index=False)
```

```python
shuffled_balanced_data.drop("attack_cat" , axis=1 , inplace=True)
shuffled_balanced_data.head()
```

```python
shuffled_balanced_data.info()
```

```python
shuffled_balanced_data["proto"].unique()
```

```python
shuffled_balanced_data["state"].unique()
```

```python
shuffled_balanced_data["service"].unique()
```

# **Scalling Using Log Scale**

```python
X = shuffled_balanced_data.drop("label",axis=1)
y = shuffled_balanced_data["label"]
```

```python
X.columns
```

```python
numeric_cols = X.select_dtypes(include=[np.number]).columns
numeric_cols
```

```python
X[numeric_cols] = np.log1p(X[numeric_cols])
```

```python
X.head(3)
```

```python
X["is_sm_ips_ports"].unique()
```

# **Encoding Using get_dummies**

```python
X = pd.get_dummies( X , dtype=int )
```

```python
X.head()
```

# **Splitting Data**

```python
from sklearn.model_selection import train_test_split
```

```python
X_train , X_test , y_train , y_test = train_test_split(X , y , random_state = 42 , stratify = y , test_size = .2)
```

```python
print(X_train.shape)
print(X_test.shape)
```

```python
X_train.describe()
```

# **Model Trainning**

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import AdaBoostClassifier
from sklearn.svm import SVC
```

```python
DT = DecisionTreeClassifier(random_state=42)
RF = RandomForestClassifier(n_estimators=50, random_state=42)
LR = LogisticRegression(max_iter=1000)
```

```python
from sklearn.ensemble import VotingClassifier
```

```python
ensemble_model = VotingClassifier(
    estimators=[
        ('lr', LR),
        ('dt', DT), 
        ('rf', RF), 
    ],
    voting='hard' 
)
```

# **Evaluation**

```python
from sklearn.metrics import classification_report, accuracy_score , confusion_matrix
```

```python
ensemble_model.fit(X_train, y_train)
y_pred = ensemble_model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

```python
DT.fit(X_train, y_train)
y_pred = DT.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

```python
RF.fit(X_train, y_train)
y_pred = RF.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

```python
LR.fit(X_train, y_train)
y_pred = LR.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```
